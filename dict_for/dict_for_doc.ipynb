Here’s a simple example of using a **for loop with a Python dictionary**:

### 🔹 Basic Example

```python
student = {
    "name": "Rahul",
    "age": 20,
    "course": "Computer Science"
}

# Loop through keys
for key in student:
    print(key)
```

### 🔹 Loop through values

```python
for value in student.values():
    print(value)
```

### 🔹 Loop through key-value pairs (most useful)

```python
for key, value in student.items():
    print(key, ":", value)
```

### 🔹 Example Output

```
name : Rahul
age : 20
course : Computer Science
```

---

### 🔹 Practical Example

```python
marks = {
    "Math": 85,
    "Science": 90,
    "English": 78
}

for subject, score in marks.items():
    if score > 80:
        print(subject, "is a strong subject")
```
