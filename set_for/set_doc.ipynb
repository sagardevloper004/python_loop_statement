# 📘 Python Set Documentation

## 🔹 What is a Set?

A **set** is a built-in Python data type used to store multiple items in a single variable.

* Unordered ❌ (no fixed position)
* Mutable ✅ (you can add/remove items)
* No duplicate values ❌

---

## 🔹 Creating a Set

```python
# Basic set
fruits = {"apple", "banana", "cherry"}

# Set with mixed data types
data = {"Rahul", 25, 5.8, True}

# Creating an empty set (important!)
empty_set = set()   # NOT {}
```

---

## 🔹 Accessing Elements

Sets are **unordered**, so you cannot access items using index.

```python
fruits = {"apple", "banana", "cherry"}

for fruit in fruits:
    print(fruit)
```

---

## 🔹 Adding Elements

```python
fruits = {"apple", "banana"}

fruits.add("cherry")
print(fruits)
```

---

## 🔹 Adding Multiple Elements

```python
fruits = {"apple", "banana"}

fruits.update(["cherry", "mango"])
print(fruits)
```

---

## 🔹 Removing Elements

```python
fruits = {"apple", "banana", "cherry"}

fruits.remove("banana")   # error if not found
fruits.discard("apple")   # no error if not found
```

---

## 🔹 Set Operations

### Union (combine sets)

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a | b)   # {1, 2, 3, 4, 5}
```

### Intersection (common elements)

```python
print(a & b)   # {3}
```

### Difference

```python
print(a - b)   # {1, 2}
```

---

## 🔹 Set Length

```python
numbers = {1, 2, 3, 4}
print(len(numbers))   # 4
```

---

## 🔹 Checking Membership

```python
fruits = {"apple", "banana"}

print("apple" in fruits)   # True
```

---

## 🔹 Set Methods

```python
s = {1, 2, 3}

s.add(4)
s.remove(2)
s.clear()   # removes all elements
```

---

## 🔹 Frozen Set (Immutable Set)

```python
fs = frozenset([1, 2, 3])
# fs.add(4) ❌ error (immutable)
```

---

## 🔹 Example Program

```python
students = {"Rahul", "Anita", "Vikas", "Rahul"}

print("Unique students:", students)

for name in students:
    print(name)
```

---

## 🔹 When to Use Set?

Use sets when:

* You need **unique values only**
* You want **fast membership testing**
* You perform **mathematical set operations**
