# 📘 Python Tuple Documentation

## 🔹 What is a Tuple?

A **tuple** is a built-in Python data type used to store multiple items in a single variable.

* Ordered ✅
* Immutable (cannot be changed) ✅
* Allows duplicate values ✅

---

## 🔹 Creating a Tuple

```python
# Basic tuple
fruits = ("apple", "banana", "cherry")

# Tuple without parentheses (packing)
numbers = 1, 2, 3, 4

# Mixed data types
data = ("Rahul", 25, 5.8, True)
```

---

## 🔹 Accessing Elements

```python
fruits = ("apple", "banana", "cherry")

print(fruits[0])   # apple
print(fruits[-1])  # cherry
```

---

## 🔹 Slicing a Tuple

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])   # (20, 30, 40)
```

---

## 🔹 Tuple is Immutable

```python
fruits = ("apple", "banana", "cherry")

# ❌ This will cause error
# fruits[0] = "mango"
```

---

## 🔹 Looping Through a Tuple

```python
colors = ("red", "green", "blue")

for color in colors:
    print(color)
```

---

## 🔹 Tuple Length

```python
numbers = (1, 2, 3, 4)

print(len(numbers))   # 4
```

---

## 🔹 Tuple with One Element

```python
single = ("apple",)   # comma is required
```

---

## 🔹 Tuple Unpacking

```python
person = ("Rahul", 20, "India")

name, age, country = person

print(name)    # Rahul
print(age)     # 20
print(country) # India
```

---

## 🔹 Nested Tuple

```python
nested = ((1, 2), (3, 4), (5, 6))

print(nested[1])      # (3, 4)
print(nested[1][0])   # 3
```

---

## 🔹 Tuple Methods

```python
numbers = (1, 2, 2, 3, 4)

print(numbers.count(2))  # 2
print(numbers.index(3))  # 3
```

---

## 🔹 When to Use Tuple?

Use tuples when:

* Data should **not change**
* You want **faster performance than lists**
* You need to use data as a **key in dictionaries**

---

## 🔹 Example Program

```python
students = (
    ("Rahul", 85),
    ("Anita", 90),
    ("Vikas", 78)
)

for name, marks in students:
    if marks > 80:
        print(name, "scored above 80")
```

---

## 🔹 Summary

| Feature    | Tuple |
| ---------- | ----- |
| Ordered    | Yes   |
| Mutable    | No    |
| Duplicates | Yes   |
| Indexed    | Yes   |
